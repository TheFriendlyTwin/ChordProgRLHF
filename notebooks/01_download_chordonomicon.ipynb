```python
from huggingface_hub import hf_hub_download
import pandas as pd
from pathlib import Path

REPO_ID = "ailsntua/Chordonomicon"
FILENAME = "chordonomicon_v2.csv"

raw_dir = Path("dataset/chordonomicon")
raw_dir.mkdir(parents=True, exist_ok=True)

local_path = hf_hub_download(
    repo_id=REPO_ID,
    filename=FILENAME,
    repo_type="dataset",
    local_dir=raw_dir
)

df = pd.read_csv(local_path)
```

```python
df
```

We'll only need the 'chords' column to compute transition probabilities.

**Note**: Because Chordonomicon does not provide an explicit musical-key annotation, the initial implementation will use a key-agnostic chord-generation policy trained directly on observed chord-transition patterns. This allows the model to learn plausible harmonic sequences without requiring key labels. Key information may be incorporated in future work by inferring the key from chord content or by conditioning the policy on user-selected keys.
